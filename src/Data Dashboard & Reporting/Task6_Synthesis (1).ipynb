## Task 6: Model Synthesis, Error Analysis, and Reporting


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
import warnings

warnings.filterwarnings('ignore')
sns.set_theme(style='whitegrid')
```

```
def plot_annotated_cm(y_true, y_pred, model_name, labels=[0, 1]):
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    cm_norm = confusion_matrix(y_true, y_pred, labels=labels, normalize='true')

    tn, fp, fn, tp = cm.ravel()
    fnr = fn / (fn + tp) if (fn + tp) > 0 else 0

    plt.figure(figsize=(6, 5))
    sns.heatmap(cm_norm, annot=True, fmt='.2%', cmap='Blues')
    plt.title(f'{model_name} Confusion Matrix\nCritical FNR: {fnr:.1%}')
    plt.show()
```

```
def plot_residual_analysis(df_results, model_name):
    fig, axes = plt.subplots(1, 3, figsize=(20, 6))

    sns.scatterplot(data=df_results, x='y_pred', y='Residuals', alpha=0.5, ax=axes[0])
    axes[0].axhline(0, color='red', linestyle='--')
    axes[0].set_title('Residuals vs Predicted')

    sns.boxplot(data=df_results, x='season', y='Residuals', ax=axes[1])
    axes[1].axhline(0, color='red', linestyle='--')
    axes[1].set_title('Residuals by Season')

    sns.boxplot(data=df_results, x='district', y='Residuals', ax=axes[2])
    axes[2].axhline(0, color='red', linestyle='--')
    axes[2].set_title('Residuals by District')

    plt.tight_layout()
    plt.show()
```
