```python
import lightgbm as lgb
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, roc_auc_score
import shap

# Import task 1 splitting module
import splitting as sp

df = pd.read_parquet("prepared_dataset.parquet")

horizons = [1, 2, 3]
results = []

# Part 1: Full CV Evaluation across Horizons
for h in horizons:
  print(f"\n================ Running Horizon: {h} ================")

  for fold_num in range(5):
    fold = sp.get_fold(
        df, fold=fold_num, horizon=h, task="clf", station_encoding="label"
    )

    X_train, y_train = fold["X_train"], fold["y_train"]
    X_test, y_test = fold["X_test"], fold["y_test"]
    class_weight = fold.get("class_weight", "balanced")

    lgb_model = lgb.LGBMClassifier(
        n_estimators=300,
        learning_rate=0.05,
        class_weight=class_weight,
        random_state=42,
        n_jobs=-1,
    )
    lgb_model.fit(X_train, y_train)
    lgb_preds_prob = lgb_model.predict_proba(X_test)[:, 1]
    lgb_preds = (lgb_preds_prob >= 0.5).astype(int)

    lgb_auc = roc_auc_score(y_test, lgb_preds_prob)
    lgb_f1 = f1_score(y_test, lgb_preds, zero_division=0)

    rf_model = RandomForestClassifier(
        n_estimators=200,
        max_depth=12,
        class_weight=class_weight,
        random_state=42,
        n_jobs=-1,
    )
    rf_model.fit(X_train, y_train)
    rf_preds_prob = rf_model.predict_proba(X_test)[:, 1]
    rf_preds = (rf_preds_prob >= 0.5).astype(int)

    rf_auc = roc_auc_score(y_test, rf_preds_prob)
    rf_f1 = f1_score(y_test, rf_preds, zero_division=0)

    results.append({
        "horizon": h,
        "fold": fold_num + 1,
        "lgb_auc": lgb_auc,
        "lgb_f1": lgb_f1,
        "rf_auc": rf_auc,
        "rf_f1": rf_f1,
    })
    print(
        f"Fold {fold_num+1} (h={h}) -> LGB AUC: {lgb_auc:.4f} | RF AUC:"
        f" {rf_auc:.4f}"
    )

# Save exact evaluation metrics
res_df = pd.DataFrame(results)
res_df.to_csv("task4_lgb_rf_metrics.csv", index=False)
print("\nMetrics summary saved to task4_lgb_rf_metrics.csv")

# Part 2: Corrected SHAP Analysis
print("\nGenerating Verified SHAP explanations for Horizon 1 (Fold 5)...")

fold_shap = sp.get_fold(
    df, fold=4, horizon=1, task="clf", station_encoding="label"
)

lgb_model_t1 = lgb.LGBMClassifier(
    n_estimators=300,
    learning_rate=0.05,
    class_weight=fold_shap.get("class_weight", "balanced"),
    random_state=42,
    n_jobs=-1,
)
lgb_model_t1.fit(fold_shap["X_train"], fold_shap["y_train"])

explainer = shap.TreeExplainer(lgb_model_t1)
shap_values = explainer.shap_values(fold_shap["X_test"])

if isinstance(shap_values, list):
  shap_vals_target = shap_values[1]
else:
  shap_vals_target = shap_values

plt.figure(figsize=(10, 6))
shap.summary_plot(
    shap_vals_target,
    fold_shap["X_test"],
    show=False,
    max_display=20,
)
plt.title(
    "Verified LightGBM Feature Impact (SHAP Summary Plot) - Horizon 1",
    pad=20,
    weight="bold",
)
plt.tight_layout()
plt.savefig("shap_summary_plot.png", dpi=300)
plt.close()

plt.figure(figsize=(10, 6))
shap.summary_plot(
    shap_vals_target,
    fold_shap["X_test"],
    plot_type="bar",
    show=False,
    max_display=20,
)
plt.title(
    "Verified LightGBM Global Feature Importance - Horizon 1",
    pad=20,
    weight="bold",
)
plt.tight_layout()
plt.savefig("shap_bar_importance.png", dpi=300)
plt.close()

print(
    "Corrected SHAP plots saved successfully: shap_summary_plot.png &"
    " shap_bar_importance.png"
)
```
